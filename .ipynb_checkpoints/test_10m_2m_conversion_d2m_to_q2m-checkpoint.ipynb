```python
import xarray as xr
import numpy as np
```

```python
PATH = '/glade/campaign/cgd/oce/people/rita/ERA5/1940/'


ds_q10 = xr.open_dataset(PATH + 'q_10.1940.nc.original_time').rename({'latitude':'lat', 'longitude':'lon'})
ds_q2 = xr.open_dataset(PATH + 'Q2M_1940.nc')
```

```python
(ds_q10.q_10.isel(time=0) - ds_q2.Q2M.isel(time=0)).plot(vmin = -0.001, vmax = 0.001, cmap = 'RdBu_r')
```

```python
%%time
(ds_q10.q_10.mean('time') - ds_q2.Q2M.mean('time')).plot(vmin = -0.001, vmax = 0.001, cmap = 'RdBu_r')
```

```python
PATH = '/glade/campaign/cgd/oce/people/rita/JRA55raw/fcst_surf/'

ds_q10 = xr.open_dataset(PATH + 'fcst_surf.q_10.reg_tl319.1958.nc', decode_times=False)
ds_q2 = xr.open_dataset(PATH + 'fcst_surf.051_q_2.reg_tl319.1958.nc', decode_times=False)
```
