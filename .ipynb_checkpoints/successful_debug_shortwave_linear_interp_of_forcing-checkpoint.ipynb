```python
import numpy as np
import xarray as xr
import pandas as pd
import matplotlib.pyplot as plt
```

# Plot swdn, swnet, lwdn from the output

```python
EXPERIMENT_DO = 'a.e30.A_JRA.TL319_t232.daily_SST_ERA5'
EXPERIMENT_RAW = 'a.e30.A_JRA55raw.TL319_t232.daily_SST_ERA5'
EXPERIMENT_ERA = 'a.e30.A_ERA5.TL319_t232.daily_SST_ERA5'

PATH_DO = '/glade/campaign/cgd/oce/people/rita/CESM_OUTPUT/' + EXPERIMENT_DO + '/1959/'
filename_atm_swnet_do = EXPERIMENT_DO + '.1959.atm_swnet.nc'
filename_atm_swdn_do = EXPERIMENT_DO + '.1959.atm_swdn.nc'
filename_atm_lwdn_do = EXPERIMENT_DO + '.1959.atm_lwdn.nc'

PATH_RAW = '/glade/campaign/cgd/oce/people/rita/CESM_OUTPUT/' + EXPERIMENT_RAW + '/1959/'
filename_atm_swnet_raw = EXPERIMENT_RAW + '.1959.atm_swnet.nc'
filename_atm_swdn_raw = EXPERIMENT_RAW + '.1959.atm_swdn.nc'
filename_atm_lwdn_raw = EXPERIMENT_RAW + '.1959.atm_lwdn.nc'


PATH_ERA = '/glade/campaign/cgd/oce/people/rita/CESM_OUTPUT/' + EXPERIMENT_ERA + '/1959/'
filename_atm_swnet_era = EXPERIMENT_ERA + '.1959.atm_swnet.nc'
filename_atm_swdn_era = EXPERIMENT_ERA + '.1959.atm_swdn.nc'
filename_atm_lwdn_era = EXPERIMENT_ERA + '.1959.atm_lwdn.nc'
```

```python
atm_swnet_do = xr.open_dataset(PATH_DO + filename_atm_swnet_do)
atm_swdn_do = xr.open_dataset(PATH_DO + filename_atm_swdn_do)
atm_lwdn_do = xr.open_dataset(PATH_DO + filename_atm_lwdn_do)

atm_swnet_raw = xr.open_dataset(PATH_RAW + filename_atm_swnet_raw)
atm_swdn_raw = xr.open_dataset(PATH_RAW + filename_atm_swdn_raw)
atm_lwdn_raw = xr.open_dataset(PATH_RAW + filename_atm_lwdn_raw)

atm_swnet_era = xr.open_dataset(PATH_ERA + filename_atm_swnet_era)
atm_swdn_era = xr.open_dataset(PATH_ERA + filename_atm_swdn_era)
atm_lwdn_era = xr.open_dataset(PATH_ERA + filename_atm_lwdn_era)
```

```python
# Reading latitude from one of the output files (to calculate weights)

PATH_out = '/glade/derecho/scratch/mmarkina/archive/' + EXPERIMENT_DO + '/cpl/hist/'
filename_out = EXPERIMENT_DO + '.cpl.hi.1959-01-01-21600.nc'

ds_out = xr.open_dataset(PATH_out + filename_out)

lat_out = ds_out.atmImp_lat.isel(time=0, drop = True)

weights = np.cos(np.deg2rad(lat_out))
weights = weights / weights.sum() 

weights.plot()
```

# Calculate mean for every timestep

```python
print('JRA-do')
swdn_do_all = [] 
lwdn_do_all = [] 
swnet_do_all = [] 

for itime in range(0, len(atm_swdn_do.time)):

    if itime%200==0:
        print(itime)

    var_name = 'atmImp_Faxa_swnet'
    out_w_mean = atm_swnet_do[var_name].isel(time = itime, drop = True).weighted(weights).mean(dim=["atmImp_nx", "atmImp_ny"]).values
    swnet_do_all.append(out_w_mean)
    
    var_name = 'atmImp_Faxa_swdn'
    out_w_mean = atm_swdn_do[var_name].isel(time = itime, drop = True).weighted(weights).mean(dim=["atmImp_nx", "atmImp_ny"]).values
    swdn_do_all.append(out_w_mean)
    
    var_name = 'atmImp_Faxa_lwdn'
    out_w_mean = atm_lwdn_do[var_name].isel(time = itime, drop = True).weighted(weights).mean(dim=["atmImp_nx", "atmImp_ny"]).values
    lwdn_do_all.append(out_w_mean)
```

```python
# Reading latitude from one of the output files (to calculate weights)

# (in principle, this is the same as JRA-do weights - just in case
PATH_out = '/glade/derecho/scratch/mmarkina/archive/' + EXPERIMENT_RAW + '/cpl/hist/'
filename_out = EXPERIMENT_RAW + '.cpl.hi.1959-01-01-21600.nc'

ds_out = xr.open_dataset(PATH_out + filename_out)

lat_out = ds_out.atmImp_lat.isel(time=0, drop = True)

weights_raw = np.cos(np.deg2rad(lat_out))
weights_raw = weights_raw / weights_raw.sum() 

weights_raw.plot()
```

```python
print('JRA-raw')
swdn_raw_all = [] 
lwdn_raw_all = [] 
swnet_raw_all = [] 

for itime in range(0, len(atm_swdn_raw.time)):

    if itime%200==0:
        print(itime)

    var_name = 'atmImp_Faxa_swnet'
    out_w_mean = atm_swnet_raw[var_name].isel(time = itime, drop = True).weighted(weights_raw).mean(dim=["atmImp_nx", "atmImp_ny"]).values
    swnet_raw_all.append(out_w_mean)
    
    var_name = 'atmImp_Faxa_swdn'
    out_w_mean = atm_swdn_raw[var_name].isel(time = itime, drop = True).weighted(weights_raw).mean(dim=["atmImp_nx", "atmImp_ny"]).values
    swdn_raw_all.append(out_w_mean)
    
    var_name = 'atmImp_Faxa_lwdn'
    out_w_mean = atm_lwdn_raw[var_name].isel(time = itime, drop = True).weighted(weights_raw).mean(dim=["atmImp_nx", "atmImp_ny"]).values
    lwdn_raw_all.append(out_w_mean)
```

```python
# Reading latitude from one of the output files (to calculate weights)

PATH_out = '/glade/derecho/scratch/mmarkina/archive/' + EXPERIMENT_ERA + '/cpl/hist/'
filename_out = EXPERIMENT_ERA + '.cpl.hi.1959-01-01-21600.nc'

ds_out = xr.open_dataset(PATH_out + filename_out)

lat_out = ds_out.atmImp_lat.isel(time=0, drop = True)

weights_era = np.cos(np.deg2rad(lat_out))
weights_era = weights_era / weights_era.sum() 

weights_era.plot()
```

```python
print('ERA')
swdn_era_all = [] 
lwdn_era_all = [] 
swnet_era_all = [] 

for itime in range(0, len(atm_swdn_raw.time)):

    if itime%200==0:
        print(itime)

    var_name = 'atmImp_Faxa_swnet'
    out_w_mean = atm_swnet_era[var_name].isel(time = itime, drop = True).weighted(weights_era).mean(dim=["atmImp_nx", "atmImp_ny"]).values
    swnet_era_all.append(out_w_mean)
    
    var_name = 'atmImp_Faxa_swdn'
    out_w_mean = atm_swdn_era[var_name].isel(time = itime, drop = True).weighted(weights_era).mean(dim=["atmImp_nx", "atmImp_ny"]).values
    swdn_era_all.append(out_w_mean)
    
    var_name = 'atmImp_Faxa_lwdn'
    out_w_mean = atm_lwdn_era[var_name].isel(time = itime, drop = True).weighted(weights_era).mean(dim=["atmImp_nx", "atmImp_ny"]).values
    lwdn_era_all.append(out_w_mean)
```

```python
swnet_out_do = xr.DataArray(
    np.array(swnet_do_all),
    dims=['time'],
    coords={'time': pd.to_datetime([str(t) for t in atm_swnet_do.time.values])})

swdn_out_do = xr.DataArray(
    np.array(swdn_do_all),
    dims=['time'],
    coords={'time': pd.to_datetime([str(t) for t in atm_swdn_do.time.values])})

lwdn_out_do = xr.DataArray(
    np.array(lwdn_do_all),
    dims=['time'],
    coords={'time': pd.to_datetime([str(t) for t in atm_swnet_do.time.values])})



swnet_out_raw = xr.DataArray(
    np.array(swnet_raw_all),
    dims=['time'],
    coords={'time': pd.to_datetime([str(t) for t in atm_swnet_raw.time.values])})

swdn_out_raw = xr.DataArray(
    np.array(swdn_raw_all),
    dims=['time'],
    coords={'time': pd.to_datetime([str(t) for t in atm_swdn_raw.time.values])})

lwdn_out_raw = xr.DataArray(
    np.array(lwdn_raw_all),
    dims=['time'],
    coords={'time': pd.to_datetime([str(t) for t in atm_swnet_raw.time.values])})


swnet_out_era = xr.DataArray(
    np.array(swnet_era_all),
    dims=['time'],
    coords={'time': pd.to_datetime([str(t) for t in atm_swnet_era.time.values])})

swdn_out_era = xr.DataArray(
    np.array(swdn_era_all),
    dims=['time'],
    coords={'time': pd.to_datetime([str(t) for t in atm_swdn_era.time.values])})

lwdn_out_era = xr.DataArray(
    np.array(lwdn_era_all),
    dims=['time'],
    coords={'time': pd.to_datetime([str(t) for t in atm_swnet_era.time.values])})
```

# Plot the forcing timeseries and mean

```python
PATH_forcing_do = '/glade/campaign/cesm/cesmdata/inputdata/ocn/jra55/v1.5_noleap/'
filename_forcing_do = 'JRA.v1.5.swdn.TL319.1959.210504.nc'

ds_forcing_do = xr.open_dataset(PATH_forcing_do + filename_forcing_do)

lat_forcing_do = ds_forcing_do.latitude
lon_forcing_do = ds_forcing_do.longitude

weights_1d = np.cos(np.deg2rad(lat_forcing_do))
weights_2d = np.outer(weights_1d, np.ones(len(lon_forcing_do)))
weights_2d = weights_2d / np.sum(weights_2d)


weights_do = xr.DataArray(
    weights_2d,
    dims=['latitude', 'longitude'],
    coords={'latitude': lat_forcing_do, 'longitude': lon_forcing_do})

del weights_2d, weights_1d
```

```python
PATH_forcing_raw = '/glade/campaign/cgd/oce/people/rita/JRA55raw/fcst_phy2m/'
filename_forcing_raw = 'fcst_phy2m.204_swdn.reg_tl319.1959.nc'

ds_forcing_raw = xr.open_dataset(PATH_forcing_raw + filename_forcing_raw, decode_times=False)


lat_forcing_raw = ds_forcing_raw.latitude
lon_forcing_raw = ds_forcing_raw.longitude

weights_1d = np.cos(np.deg2rad(lat_forcing_raw))
weights_2d = np.outer(weights_1d, np.ones(len(lon_forcing_raw)))
weights_2d = weights_2d / np.sum(weights_2d)


weights_raw = xr.DataArray(
    weights_2d,
    dims=['latitude', 'longitude'],
    coords={'latitude': lat_forcing_raw, 'longitude': lon_forcing_raw})

del weights_2d, weights_1d
```

```python
PATH_forcing_era = '/glade/campaign/cgd/oce/people/rita/ERA5/1959/'
filename_forcing_era = 'swdn.1959.nc'

ds_forcing_era = xr.open_dataset(PATH_forcing_era + filename_forcing_era, decode_times=False)


lat_forcing_era = ds_forcing_era.latitude
lon_forcing_era = ds_forcing_era.longitude

weights_1d = np.cos(np.deg2rad(lat_forcing_era))
weights_2d = np.outer(weights_1d, np.ones(len(lon_forcing_era)))
weights_2d = weights_2d / np.sum(weights_2d)


weights_era = xr.DataArray(
    weights_2d,
    dims=['latitude', 'longitude'],
    coords={'latitude': lat_forcing_era, 'longitude': lon_forcing_era})

del weights_2d, weights_1d
```

```python
swdn_forcing_do = [] 

for itime in range(0, len(ds_forcing_do.time)):

    if itime%200==0:
        print(itime)
    
    var_name = 'swdn'
    out_w_mean = ds_forcing_do[var_name].isel(time = itime, drop = True).weighted(weights_do).mean(dim=["latitude", "longitude"])
    swdn_forcing_do.append(out_w_mean)
```

```python
swdn_forcing_raw = [] 

for itime in range(0, len(ds_forcing_raw.time)):

    if itime%200==0:
        print(itime)
    
    var_name = 'swdn'
    out_w_mean = ds_forcing_raw[var_name].isel(time = itime, drop = True).weighted(weights_raw).mean(dim=["latitude", "longitude"])
    swdn_forcing_raw.append(out_w_mean)
```

```python
swdn_forcing_era = [] 

for itime in range(0, len(ds_forcing_era.time)):

    if itime%200==0:
        print(itime)
    
    var_name = 'swdn'
    out_w_mean = ds_forcing_era[var_name].isel(time = itime, drop = True).weighted(weights_era).mean(dim=["latitude", "longitude"])
    swdn_forcing_era.append(out_w_mean)
```

```python
if len(ds_forcing_do.time) == len(ds_forcing_raw.time):
    ds_forcing_raw['time'] = ds_forcing_do.time.copy()
```

```python
swdn_forcing_do = xr.DataArray(
    swdn_forcing_do,
    dims=['time'],
    coords={'time': pd.to_datetime([str(t) for t in ds_forcing_do.time.values])})

swdn_forcing_raw = xr.DataArray(
    swdn_forcing_raw,
    dims=['time'],
    coords={'time': pd.to_datetime([str(t) for t in ds_forcing_raw.time.values])})

swdn_forcing_era = xr.DataArray(
    swdn_forcing_era,
    dims=['time'],
    coords={'time': pd.to_datetime([str(t) for t in ds_forcing_era.time.values])})
```
